# K-Means Clustering

## What is Clustering?

Clustering is the task of grouping a set of objects in such a way that objects in the same group (called a cluster) are more similar to each other than to those in other groups (clusters).

## What is K-Means Clustering?

K-Means is a popular algorithm for clustering analysis in data mining. The algorithm works as follows:

1. Select the number of clusters, k.
2. Randomly initialize k centroids.
3. Assign each data point to the cluster with the closest centroid.
4. Recompute the centroids of the clusters.

## The goal is to:
- Minimize the sum of squared distances between each data point and its assigned centroid.

## How to choose the number of clusters, k?

1. Elbow Method
2. Silhouette Score
3. WCSS (Within Cluster Sum of Squares)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Generate some random data
np.random.seed(0)
X = np.random.rand(100, 2)

# Apply K-Means clustering
kmeans = KMeans(n_clusters=3)
kmeans.fit(X)
```

```python
# Plot the data points and the centroids
plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', marker='x')
plt.show()
```

## Implementation of K-Means Clustering

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

```python
# create custom dataset
X, y = make_blobs(n_samples=500, centers=3, n_features=3, random_state=42)
# initialize k-means
kmeans = KMeans(n_clusters=3, random_state=42)
# fit the model
kmeans.fit(X)
# get the cluster centers
centroids = kmeans.cluster_centers_
# plot the data points and the centroids
plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_)
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='x')
plt.show()
```
